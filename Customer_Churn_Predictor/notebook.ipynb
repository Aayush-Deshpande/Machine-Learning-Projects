```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as ptl
```

```python
df = pd.read_csv("Churn_Modelling.csv", index_col=0)
```

```python
df.duplicated().sum()
```

```python
df.head()
```

```python
df.drop(columns=['CustomerId', 'Surname'], axis=1, inplace=True)
```

```python
df.head()
```

```python
X = df.drop(columns=['Exited'], axis=1)
y = df['Exited']
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer

cat_col = [col for col in X.columns if X[col].dtype == 'O']
num_col = [col for col in X.columns if X[col].dtype != 'O']

ct = ColumnTransformer([
    ("OneHotEncoding", OneHotEncoder(drop='first'), cat_col),
    ("StandardScaler", StandardScaler(), num_col)
])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

X_train = ct.fit_transform(X_train)
X_test = ct.transform(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred, target_names=['Stayed', 'Exited']))
```
